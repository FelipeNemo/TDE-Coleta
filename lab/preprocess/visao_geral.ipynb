```python
import os
import pandas as pd

base_dir = "../../data/processed"

# 1) Gerar lista de arquivos que começam com "votacoes_" e terminam com ".csv"
arquivos = [
    os.path.join(base_dir, nome)
    for nome in os.listdir(base_dir)
    if nome.startswith("votacoes_") and nome.endswith(".csv")
]

# garantir que fiquem em ordem (votacoes_2020-01, 2020-02, ..., 2025-10, etc.)
arquivos = sorted(arquivos)

# colunas obrigatórias
colunas_obrigatorias = [
    "id",
    "uri",
    "data",
    "dataHoraRegistro",
    "siglaOrgao",
    "uriOrgao",
    "uriEvento",
    "proposicaoObjeto",
    "uriProposicaoObjeto",
    "descricao",
    "aprovacao",
]

# 2) Ler todos, mas só manter quem tem todas as colunas obrigatórias
dfs = []
for f in arquivos:
    try:
        df = pd.read_csv(f)
    except pd.errors.EmptyDataError:
        print(f"[AVISO] {f} vazio/sem colunas, ignorando.")
        continue

    if set(colunas_obrigatorias).issubset(df.columns):
        dfs.append(df)
    else:
        print(f"[AVISO] {f} não tem todas as colunas obrigatórias, ignorando.")

# 3) Concatenar tudo em um DataFrame só
votacoes = pd.concat(dfs, ignore_index=True)

# 4) Ordenar pela coluna "data"
# como está no formato YYYY-MM-DD, dá para ordenar como string mesmo
votacoes = votacoes.sort_values("data").reset_index(drop=True)

# 5) Salvar o arquivão
votacoes.to_csv("../../data/processed/votacoes_2020-02_2025-10_todas.csv", index=False)
```
